|-----------|------------------|--------------------|--------------------|

```python
#entrda
nombre = input("Ingrese su nombre: ").upper().strip()
#proceso
try:
  numero = int(input("Ingrese un numero: "))
  if numero % 2 == 0:
    print(nombre, "el numero es par. ")
  else:
    print(nombre, "el numero es impar. ")

except ValueError:
 #salida
  print("Solo se permiten numeros enteros")
```
